```python
# import pandas for data manipulation
import pandas as pd

# import plotly express for interactive visualizations
import plotly.express as px

print("Libraries loaded")
```

```python
# load dengue dataset
data = pd.read_csv("../data/raw/dengue_dataset_original.csv")

# display first 5 rows
data.head()
```

```python
# summary statistics - to find abnormilities
data.describe().transpose()
```

```python
# print the number of rows and cols
data.shape
```

```python
# check missing values of each column
data.isnull().sum()
```

```python
import plotly.express as px

#histogram - helps identify extreme case counts
cases_hist = px.histogram(
    data,
    x="cases",
    title="Distribution of Dengue Cases"
)

cases_hist.show()
```

```python
#detect extreme case spikes
cases_box = px.box(
    data,
    x="cases",
    title="Box Plot of Dengue Cases"
)

cases_box.show()
```

```python
#detect extreme rain spikes 
rain_box = px.box(
    data,
    x="avg_precipitation_sum",
    title="Box Plot of Rainfall"
)

rain_box.show()
```

```python
#shows the relationship between rainfall and dengue cases
rain_cases = px.scatter(
    data,
    x="avg_precipitation_sum",
    y="cases",
    title="Rainfall vs Dengue Cases"
)

rain_cases.show()
```

```python
#shows relationship between temperature and dengue cases
temp_cases = px.scatter(
    data,
    x="avg_temperature_mean",
    y="cases",
    title="Temperature vs Dengue Cases"
)

temp_cases.show()
```

```python
#shows how dengue evolved over time
area_data = data[data["moh_area"] == data["moh_area"].unique()[0]]

trend_plot = px.line(
    area_data,
    x="week",
    y="cases",
    title="Weekly Dengue Cases Trend"
)

trend_plot.show()
```

```python
#shows how temperature affected the dengue case counts over time
temp_hist = px.histogram(
    data,
    x="avg_temperature_mean",
    color="moh_area",
    title="Temperature Distribution by MOH Area",
    barmode="overlay"
)

temp_hist.show()
```
